# 4. Image classification with transfer learning
Use pretrained MobileNetV2 in TensorFlow/Keras to predict ImageNet categories. This avoids training a CNN from scratch and is suitable for a live demo.

```python
import tensorflow as tf, numpy as np, matplotlib.pyplot as plt
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input, decode_predictions
from PIL import Image
import requests
from io import BytesIO
print("TensorFlow:", tf.__version__, "| GPU devices:", tf.config.list_physical_devices("GPU"))
```

```python
model = MobileNetV2(weights="imagenet")
url = "https://storage.googleapis.com/download.tensorflow.org/example_images/YellowLabradorLooking_new.jpg"
image = Image.open(BytesIO(requests.get(url, timeout=30).content)).convert("RGB").resize((224,224))
arr = np.array(image, dtype=np.float32)[None,...]
probs = model.predict(preprocess_input(arr), verbose=0)
plt.figure(figsize=(4,4)); plt.imshow(image); plt.axis("off"); plt.show()
for _, label, score in decode_predictions(probs, top=5)[0]:
    print(f"{label}: {score:.1%}")
```

## Why transfer learning?
A pretrained network has already learned useful visual patterns. We can reuse it for predictions immediately, or replace its final classification layer and fine-tune on our own labeled images.

**Expected output:** top-5 ImageNet labels and probabilities. The demo is a general classifier, not a dedicated cat-vs-dog model.
