````python
from ultralytics import YOLO
import torch, matplotlib.pyplot as plt
from PIL import Image
from IPython.display import display
print("PyTorch:", torch.__version__, "| CUDA available:", torch.cuda.is_available())
````

````python
# Pretrained COCO model; first run downloads weights.
model = YOLO("yolo11n.pt")
# Ultralytics sample image downloaded by package helper, or replace with your own image path.
from ultralytics.utils.downloads import download
sample_path = "https://ultralytics.com/images/bus.jpg"
results = model.predict(source=sample_path, conf=0.25, device=0 if torch.cuda.is_available() else "cpu", verbose=False)
annotated = results[0].plot()[..., ::-1]
plt.figure(figsize=(12,7)); plt.imshow(annotated); plt.axis("off"); plt.title("YOLO detections"); plt.show()
for b in results[0].boxes:
    cls_id=int(b.cls[0]); print(model.names[cls_id], "confidence:", round(float(b.conf[0]),3), "xyxy:", [round(float(x),1) for x in b.xyxy[0]])
````
