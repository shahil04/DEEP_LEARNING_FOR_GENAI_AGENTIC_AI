```python
import cv2, numpy as np, matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path

# Create a simple synthetic image so the notebook runs without a download.
img = np.zeros((420, 640, 3), dtype=np.uint8) + 245
cv2.rectangle(img, (70, 80), (230, 280), (40, 120, 220), -1)
cv2.circle(img, (390, 170), 80, (40, 180, 80), -1)
pts = np.array([[470,300],[560,300],[515,210]], np.int32)
cv2.fillPoly(img, [pts], (200,70,70))
plt.figure(figsize=(10,5)); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.title("Synthetic sample image"); plt.axis("off");
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5,5), 0)
edges = cv2.Canny(blur, 50, 150)
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
annotated = img.copy()
for c in contours:
    if cv2.contourArea(c) < 500: 
        continue
    x,y,w,h = cv2.boundingRect(c)
    cv2.rectangle(annotated,(x,y),(x+w,y+h),(0,0,0),3)
    cv2.putText(annotated, f"contour {int(cv2.contourArea(c))}", (x,y-8), cv2.FONT_HERSHEY_SIMPLEX, .5, (0,0,0), 2)
fig, ax = plt.subplots(1,3,figsize=(15,5))
ax[0].imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)); ax[0].set_title("Input")
ax[1].imshow(edges,cmap="gray"); ax[1].set_title("Canny edges")
ax[2].imshow(cv2.cvtColor(annotated,cv2.COLOR_BGR2RGB)); ax[2].set_title(f"Contours: {len(contours)}")
for a in ax: a.axis("off")
plt.show()
```

## Explain to students
- **Thresholding:** separates pixels using an intensity rule.
- **Canny:** highlights strong image edges.
- **Contours:** groups connected edge points into outlines.
- **Limitation:** contours do not understand semantic concepts like “car” or “person”; lighting and background can change results.

**Class activity:** change the Canny thresholds from `50, 150` to `20, 80` and compare.
