# 6. Facial-expression image classification
This demo uses a pretrained facial-expression classifier. It predicts visual expression categories (e.g., happy, sad, neutral), **not a person's true internal feelings**.

Upload a suitable face image, or provide a public image URL. Use consented sample images in class.

```python
from transformers import pipeline
from PIL import Image
from io import BytesIO
import requests, matplotlib.pyplot as plt
# First run downloads model files. Internet required.
classifier = pipeline("image-classification", model="trpakov/vit-face-expression")
```

```python
# Replace this URL with a consented face image URL, or upload a file in Colab.
image_url = "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/transformers/tasks/image_classification.png"
try:
    image = Image.open(BytesIO(requests.get(image_url, timeout=30).content)).convert("RGB")
    display(image)
    predictions = classifier(image)
    for p in predictions[:5]:
        print(f"{p['label']}: {p['score']:.1%}")
except Exception as e:
    print("The placeholder may not contain a face. Upload a face image and run classifier(your_image).")
    print(type(e).__name__, str(e)[:300])
```

## Classroom explanation
- Input: face image.
- Model: pretrained Vision Transformer classifier.
- Output: ranked expression labels and scores.
- Caveat: expressions vary by person and context; predictions can be biased and must not be treated as proof of emotion or used for high-stakes decisions.
